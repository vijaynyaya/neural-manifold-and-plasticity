```python
import numpy as np
```

```python
class RecursiveLeastSquares:
    def __init__(self, n_features, delta=1000):
        self.n = n_features
        self.theta = np.zeros((n_features, 1))  # initial weights
        self.P = np.eye(n_features) * delta     # large initial covariance

    def update(self, x_t, y_t):
        x_t = x_t.reshape(-1, 1)
        y_t = np.array([[y_t]])

        # Compute gain
        numerator = self.P @ x_t
        denominator = 1 + x_t.T @ self.P @ x_t
        K_t = numerator / denominator

        # Update theta
        error = y_t - x_t.T @ self.theta
        self.theta = self.theta + K_t @ error

        # Update P
        self.P = self.P - K_t @ x_t.T @ self.P

        return self.theta.flatten()
```
