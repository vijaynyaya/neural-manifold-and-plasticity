```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 3
g = 1.5
p = 0.5
np.random.seed(2025)
```

```python
mask = np.random.rand(N, N) < p
np.fill_diagonal(mask, False) 
w = np.random.randn(N, N) * mask
w
```

```python
im = plt.imshow(w, cmap="coolwarm")
for i, j in np.ndindex(w.shape):
    if mask[i, j]:
        im.axes.text(j, i, f"{w[i, j]:.2f}", ha='center', va='center', color='white')
plt.show()
```

```python
W_p = [
    np.where(w[i,:] != 0)[0] # non-zero weights
    for i in range(N)
]
W_p
```
